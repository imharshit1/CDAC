```python
duplo_members = df[(df["Toy Company"] == "Duplo") & (df["Member?"] == "Yes")]
duplo_members
```

```python
df['Total COGS'] = df['Total COGS'].astype(str).str.replace("$", "", regex=False).str.strip().astype(float)
```

```python
df
```

```python
currency_cols = ["Price Per Toy", "Total Line Revenue"]
for col in currency_cols:
    df[col] = df[col].astype(str).str.replace("$", "", regex=False).str.strip().astype(float)
```

```python
df
```

```python
df['Profit Percent'] = (df['Total Line Revenue'] - df['Total COGS']) / df['Total Line Revenue'] * 100
```

```python
df['Profit'] = df['Total Line Revenue'] - df['Total COGS']
```
