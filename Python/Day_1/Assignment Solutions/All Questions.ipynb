```python
#Write a program that takes a year as input from the user and checks whether it is a leap year or not.

def leap_year(year):
        
    if year%100==0:
        if year%4==0 and year%400==0:
            print(f"{year} is leap year")
    elif year%4==0:
            print(f"{year} is leap year")
        
    else:
        print(f"{year} is not a leap year ")        
def main():
    try:
        year=int(input("Enter the year:"))
    except ValueError:
        print("Enter the valid year")
        return

    leap_year(year)
    
main()
    
            
    
    
```

```python
def complete_series(n, s, next_prev, prev):
    if(len(s) >= n):
        return
    curr = prev + next_prev
    s.append(curr)
    complete_series(n, s, prev, curr)
    
def main():
    n = int(input("Enter the number please: "))
    s = [0, 1]
    complete_series(n, s, 0, 1)
    print(s)
main()
    
```
